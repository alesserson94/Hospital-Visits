```python
#Import pandas & datetime packages
import pandas as pd
import datetime as dt

#Display all values
pd.set_option('display.max_rows',None)
```

```python
hospital_stays=pd.read_csv('/Users/aaronless/Desktop/Data Analysis Porfolio Projects/Other/Hospital Stays/Original CSVs/hospital_stays.csv')
hospital_stays.head(50)
```

```python
#Identify any duplicate entries within the data
stay_duplicates=hospital_stays.duplicated()
hospital_stays[stay_duplicates].sort_values(by='stay_id')
```

```python
#Remove all duplicate values
hospital_stays.drop_duplicates().head(50)
```

```python
#Edit patient_id column to remove the prefix PT before each value
hospital_stays['patient_id']=hospital_stays['patient_id'].str.strip('PT-')
hospital_stays['patient_id']=hospital_stays['patient_id'].str.strip('pt-')
hospital_stays.head(50)
```

```python
#Edit admission_date column to ensure consistent date formatting
hospital_stays['admission_date']=pd.to_datetime(hospital_stays['admission_date']).dt.date
hospital_stays.head(50)
```

```python
#Clean discharge_date column for consistent date formatting
hospital_stays['discharge_date']=pd.to_datetime(hospital_stays['discharge_date']).dt.date
hospital_stays.head(50)
```

```python
#Inspect values in admission_type column and determine data errors
hospital_stays['admission_type'].value_counts()
```

```python
#Inspect department column to identify data inconsistencies
hospital_stays['department'].value_counts()

#Ensure proper capitalization and remove trailing whitespaces 
hospital_stays['department']=hospital_stays['department'].str.title()
hospital_stays['department']=hospital_stays['department'].str.strip()

hospital_stays['department'].value_counts()
```

```python
hospital_stays['diagnosis_description'].value_counts()

#Ensure proper capitalization across all diagnosis descriptions
hospital_stays['diagnosis_description']=hospital_stays['diagnosis_description'].str.title()
hospital_stays['diagnosis_description'].value_counts()
```

```python
hospital_stays['room_type'].value_counts()
```

```python
hospital_stays['attending_physician']=hospital_stays['attending_physician'].str.strip('Dr.')
hospital_stays.head(50)
```

```python
hospital_stays['readmission_within_30_days'].value_counts()

hospital_stays['readmission_within_30_days']=hospital_stays['readmission_within_30_days'].replace({'Y':'Yes','N':'No','y':'Yes','n':'No'})
hospital_stays['readmission_within_30_days'].value_counts()
```

```python
hospital_stays['total_charge_usd']=hospital_stays['total_charge_usd'].str.strip('$')
hospital_stays['total_charge_usd']=hospital_stays['total_charge_usd'].str.strip(',')
hospital_stays['total_charge_usd'].value_counts().head(50)
```

```python
hospital_stays.head(50)
```

```python
hospital_stays_cleaned=hospital_stays.to_csv('~/Desktop/hospital_stays_cleaned.csv')
```
